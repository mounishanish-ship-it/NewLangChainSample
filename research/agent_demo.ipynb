```python
import os
import certifi
import requests

from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
#from langchain_tools import langchain_tools
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain import hub
```

```python
from langchain.agents import create_react_agent, AgentExecutor

```

```python
#LOAD ENVIRONMENT VARIABLES
os.environ["REQUESTS_CA_BUNDLE"] = certifi.where()
load_dotenv()
#print(os.getenv("TAVILY_API_KEY"))
openai_api_key = os.getenv("OPENAI_API_KEY")
tavily_api_key = os.getenv("TAVILY_API_KEY")

```

```python
search_tool = TavilySearchResults(max_results=5)
result = search_tool.invoke("What is the latest news on AI?")

result
```

```python
# Creating LLM 

llm = ChatOpenAI(model_name="gpt-5.6-terra", temperature=1, openai_api_key=openai_api_key)
```

```python
response = llm.invoke("tell me a joke about AI")
response
```

```python
#creating a prompt with react agent

prompt = hub.pull("hwchase17/react")
```

```python
#tools

tools = [search_tool]
```

```python
prompt
```

```python
# creating a react agent

agent = create_react_agent(llm=llm, tools=tools, prompt=prompt)
```

```python
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)
```

```python
#run the agent executor with a query

response = agent_executor.invoke(
    {"input": "Find the capital of India "}

)
```
